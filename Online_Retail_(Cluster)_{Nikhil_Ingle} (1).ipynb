# **Project Summary -**

### Summary of the **Online Retail** Dataset

The **Online Retail** dataset is a comprehensive collection of transactional data, representing sales records for an e-commerce business. The dataset contains eight key columns that provide valuable insights into sales, customer behavior, and product performance. Here’s an overview of its structure and potential insights:

#### Key Data Columns:
1. **InvoiceNo**:  
   This is the unique identifier for each transaction, acting as a primary key. It helps trace individual sales and is essential for linking multiple rows that are part of the same invoice.

2. **StockCode**:  
   A unique code assigned to each product. This field enables the identification of specific products, their frequency of purchase, and their overall contribution to sales.

3. **Description**:  
   This column provides detailed names or descriptions of the products sold. This is crucial for analyzing the performance of various items and identifying high-demand or slow-moving inventory.

4. **Quantity**:  
   Indicates the number of units of a particular product purchased in a transaction. It can help calculate total sales volume and understand buying trends.

5. **InvoiceDate**:  
   The date and time when the transaction occurred. This allows for time-series analysis, seasonal trend identification, and the detection of peak sales periods.

6. **UnitPrice**:  
   The price per unit of a product. Combined with quantity, this column helps calculate the total revenue generated from each product and invoice.

7. **CustomerID**:  
   A unique identifier for each customer, enabling customer segmentation, behavior analysis, and the determination of repeat purchase trends.

8. **Country**:  
   Indicates the country in which the transaction took place, offering insights into regional sales performance and market penetration.

#### Characteristics of the Dataset:
- **Transaction Diversity**:  
  Each invoice may contain multiple products, as customers often purchase a variety of items. This creates opportunities to analyze product bundling and cross-selling strategies.

- **Timeframe**:  
  The dataset includes detailed timestamps, making it ideal for understanding sales patterns over time, including daily, weekly, and seasonal trends.

- **Geographic Scope**:  
  With sales data spanning multiple countries, the dataset allows for a comparative analysis of international markets and customer preferences.

- **Customer Insights**:  
  The presence of unique customer identifiers supports loyalty analysis, customer lifetime value calculations, and the design of targeted marketing campaigns.

#### Potential Applications:
1. **Sales Analysis**:  
   Understanding which products generate the most revenue, identifying best-sellers, and recognizing underperforming items.

2. **Customer Behavior Analysis**:  
   Examining purchasing habits, including the frequency of purchases, average basket size, and the correlation between product categories.

3. **Market Trends**:  
   Leveraging the InvoiceDate column to identify seasonal spikes in demand or time-based patterns in customer activity.

4. **Inventory Management**:  
   Using stock-level data to optimize inventory turnover, reduce holding costs, and ensure the availability of popular items.

5. **Regional Analysis**:  
   Comparing sales data across countries to tailor marketing strategies to local preferences.

#### Challenges:
- **Missing Data**:  
  Some columns, like CustomerID, may have missing values, which could limit the depth of customer-specific insights.
  
- **Outliers**:  
  Transactions with unusually high or negative quantities (e.g., returns or errors) may require cleaning for accurate analysis.

- **Data Volume**:  
  With potentially large datasets, computational efficiency and proper data handling techniques are essential.

#### Conclusion:
The **Online Retail** dataset provides a wealth of information for understanding sales performance, customer behavior, and market trends. It is highly suited for generating actionable insights in domains like retail analytics, customer relationship management, and inventory optimization. With proper cleaning and exploration, this dataset can serve as a robust foundation for driving business decisions and enhancing operational efficiency.


# Feature Engineering








````python
# creating new variable 'TotalPrice'
df['TotalPrice'] = df['Quantity'] * df['UnitPrice']
````

````python
# Creating new features from the datetime column InvoiceDate
df['date'] = df['InvoiceDate'].dt.date
df["year"]  = df["InvoiceDate"].apply(lambda x: x.year)
df['month'] = df['InvoiceDate'].apply(lambda x: x.month_name())
df['day']   = df['InvoiceDate'].apply(lambda x: x.day_name())
df['weekday'] = df['InvoiceDate'].dt.day_name()
df["hour"]  = df["InvoiceDate"].apply(lambda x: x.hour)
````

````python
# Creating a new feature 'TimeType' based on hours to define whether its Morning,Afternoon or Evening
df['TimeType'] = np.where((df["hour"]>5)&(df["hour"]<18), np.where(
                           df["hour"]<12, 'Morning','Afternoon'),'Evening')
````

````python
# InvoiceNo starting with 'C' represents cancellation
df['InvoiceNo'] = df['InvoiceNo'].astype('str')
cancellations = df[df['InvoiceNo'].str.contains('C')]
````

````python
cancellations.head()
````

````python
# Dropping cancellations from the main dataframe
df = df[~df['InvoiceNo'].str.contains('C')]
````

## ***6. ML Model Implementation***

### ML Model

#RFM model (Recency, Frequency, and Monetary value)

````python
df.columns
````

````python
# ML Model - 1 Implementation
# Calculating Recency, Frequency and Monetary.
#Set Latest date 2011-12-10 as last invoice date was 2011-12-09. This is to calculate the number of days from recent purchase
import datetime as dt

# Set the latest date for recency calculation
Latest_Date = dt.date(2011, 12, 10)

# Create RFM values for each customer
rfm_df = df.groupby('CustomerID').agg({
    'date': lambda x: (Latest_Date - x.max()).days,  # Recency: Days since last purchase
    'InvoiceNo': 'count',  # Frequency: Number of invoices
    'TotalPrice': 'sum'    # Monetary: Total amount spent
})

# Rename columns to Recency, Frequency, and Monetary
rfm_df.rename(columns={
    'date': 'Recency',
    'InvoiceNo': 'Frequency',
    'TotalPrice': 'Monetary'
}, inplace=True)

# Reset index for a clean DataFrame
rfm_df = rfm_df.reset_index()

# Display the first few rows
print(rfm_df.head())
````

````python
# data summary of Recency, Frequency, and Monetary
rfm_df.describe()
````

````python
# ploting distributions of Recency, Frequency, and Monetary
fig = plt.figure(figsize=(12,6))
c=1
for i in ['Recency' , 'Frequency', 'Monetary'] :
    plt.subplot(2, 2, c)
    plt.xlabel(i, fontsize = 12)
    sns.distplot(rfm_df[i],color="tomato")
    c = c + 1
plt.tight_layout()
````

### Calculating RFM scores.

````python
# geting Q1, Q2 and Q3
quantiles = rfm_df.quantile(q=[0.25,0.5,0.75])
quantiles = quantiles.to_dict()
quantiles
````

#### Giving R, F and M score from 1 to 4 where four is the best score and one is worse..

````python
#Functions to calculate R, F and M scoring
def Rscoring(x,p,d):
    if x <= d[p][0.25]:
        return 4
    elif x <= d[p][0.50]:
        return 3
    elif x <= d[p][0.75]:
        return 2
    else:
        return 1

def FandMscoring(x,p,d):
    if x <= d[p][0.25]:
        return 1
    elif x <= d[p][0.50]:
        return 2
    elif x <= d[p][0.75]:
        return 3
    else:
        return 4
````

````python
# Calculating and creating  R, F and M scores columns.
rfm_df['R'] = rfm_df['Recency'].apply(Rscoring, args=('Recency',quantiles,))
rfm_df['F'] = rfm_df['Frequency'].apply(FandMscoring, args=('Frequency',quantiles,))
rfm_df['M'] = rfm_df['Monetary'].apply(FandMscoring, args=('Monetary',quantiles,))
rfm_df.head()
````

````python
# Assigning and creating RFM_group column showing combined concatenated score of RFM.
rfm_df['RFM_group'] = rfm_df.R.map(str) + rfm_df.F.map(str) + rfm_df.M.map(str)

# Assigning and creating RFM_score column showing total sum of R, F and M scores.
rfm_df['RFM_score'] = rfm_df[['R', 'F', 'M']].sum(axis = 1)
rfm_df.head()
````

* RFM_group : RFM_group is combined concatenated score of R, F and M scores, where 444 represents best and 111 represents worst.
* RFM_score : RFM_score is total sum of R, F and M scores, where 12 is highest and best score and 3 is lowest and worse score.

### Silhouette analysis
Silhouette analysis is a technique used to evaluate the quality of clusters formed by a clustering algorithm. It involves calculating a silhouette coefficient for each data point in the dataset, which measures how similar that point is to its own cluster compared to other clusters.

The silhouette coefficient ranges from -1 to 1, with values closer to 1 indicating that the data point is well-matched to its cluster, and values closer to -1 indicating that the data point would be better matched to a different cluster. A value of 0 indicates that the data point is on the boundary between two clusters.

To perform silhouette analysis, one first needs to cluster the data using a clustering algorithm such as k-means. Once the clusters have been formed, the silhouette coefficient for each data point is calculated using the following formula:

$ s(i) = \frac{(b(i) - a(i))}{max(a(i), b(i))}$

where s(i) is the silhouette coefficient for the i-th data point, a(i) is the average distance between i and all other points in the same cluster, and b(i) is the average distance between i and all points in the nearest neighboring cluster (i.e., the cluster with the smallest average distance to i).

#### Determining the best value of k using silhoutte method.
The silhouette method can also be used to determine the best value of k (i.e., the number of clusters) for a given dataset. To do so, one can calculate the average silhouette coefficient for each value of k and choose the value that maximizes this metric.

````python
# input variables
rfm_df = rfm_df[['Recency', 'Frequency', 'Monetary']]

# data scaling using StandardScaler
scaler = StandardScaler()
# fit_transform
X = scaler.fit_transform(rfm_df)
````

````python
# determining the best value of k using silhoutte method
for k in range(2,11):
  model = KMeans(k)
  preds = model.fit_predict(X)
  centers = model.cluster_centers_
  # getting silhoutte score
  score = silhouette_score(X, preds)
  print("For n_clusters = {}, silhouette score is {}".format(k, score))
````
